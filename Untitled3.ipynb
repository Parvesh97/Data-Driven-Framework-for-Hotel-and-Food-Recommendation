```python
# 🔹 Step 1 — Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

# 🔹 Step 2 — Load Dataset
df = pd.read_csv("Dataset .csv")

# 🔹 Step 3 — Basic Cleaning
df.columns = df.columns.str.strip().str.replace(" ", "_")
df['Cuisines'] = df['Cuisines'].fillna("Unknown")

df['Has_Online_delivery'] = df['Has_Online_delivery'].map({'Yes':1,'No':0})
df['Has_Table_booking'] = df['Has_Table_booking'].map({'Yes':1,'No':0})

print("\nDataset Shape:", df.shape)

# =========================================================
# 🔥 LEVEL 1 TASKS
# =========================================================

print("\n================ LEVEL 1 =================")

# 🍜 Task 1 — Top Cuisines
print("\nTop 3 Cuisines:")
cuisines = df['Cuisines'].str.split(', ')
top_cuisines = cuisines.explode().value_counts().head(3)
print(top_cuisines)

print("\nPercentage of Restaurants Serving Top Cuisines:")
print((top_cuisines/len(df))*100)

# 🏙 Task 2 — City Analysis
print("\nCity with Highest Restaurants:", df['City'].value_counts().idxmax())

city_rating = df.groupby('City')['Aggregate_rating'].mean().sort_values(ascending=False)
print("\nTop Cities by Average Rating:")
print(city_rating.head())
print("\nCity with Highest Avg Rating:", city_rating.idxmax())

# 💰 Task 3 — Price Range Distribution
plt.figure()
sns.countplot(x=df['Price_range'])
plt.title("Price Range Distribution")
plt.show()

print("\nPrice Range Percentage:")
print(df['Price_range'].value_counts(normalize=True)*100)

# 🚚 Task 4 — Online Delivery
print("\nOnline Delivery Percentage:",
      df['Has_Online_delivery'].mean()*100)

print("\nAverage Ratings (Delivery vs No Delivery):")
print(df.groupby('Has_Online_delivery')['Aggregate_rating'].mean())

# =========================================================
# 🔥 LEVEL 2 TASKS
# =========================================================

print("\n================ LEVEL 2 =================")

# ⭐ Task 1 — Restaurant Ratings
plt.figure()
plt.hist(df['Aggregate_rating'], bins=20)
plt.title("Rating Distribution")
plt.show()

print("\nMost Common Rating Range:")
print(df['Aggregate_rating'].round().value_counts().head())

print("\nAverage Votes:", df['Votes'].mean())

# 🍽 Task 2 — Cuisine Combination
print("\nMost Common Cuisine Combinations:")
print(df['Cuisines'].value_counts().head(10))

print("\nTop Rated Cuisine Combinations:")
print(df.groupby('Cuisines')['Aggregate_rating']
      .mean().sort_values(ascending=False).head())

# 🗺 Task 3 — Geographic Analysis
plt.figure()
plt.scatter(df['Longitude'], df['Latitude'], alpha=0.3)
plt.title("Restaurant Locations Map")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.show()

# 🏬 Task 4 — Restaurant Chains
chains = df['Restaurant_Name'].value_counts()
chains = chains[chains > 1]

print("\nRestaurant Chains:")
print(chains.head())

print("\nTop Rated Restaurant Chains:")
print(df[df['Restaurant_Name'].isin(chains.index)]
      .groupby('Restaurant_Name')['Aggregate_rating']
      .mean().sort_values(ascending=False).head())

# =========================================================
# ⭐ BONUS — LEVEL 3 (Extra Strong Submission)
# =========================================================

print("\n================ BONUS LEVEL 3 =================")

print("\nHighest Votes Restaurant:")
print(df.loc[df['Votes'].idxmax()][['Restaurant_Name','Votes']])

print("\nLowest Votes Restaurant:")
print(df.loc[df['Votes'].idxmin()][['Restaurant_Name','Votes']])

correlation = df['Votes'].corr(df['Aggregate_rating'])
print("\nCorrelation Between Votes & Rating:", correlation)

plt.figure()
sns.scatterplot(x='Votes', y='Aggregate_rating', data=df)
plt.title("Votes vs Rating")
plt.show()
```
